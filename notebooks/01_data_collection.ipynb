# STEP 1: SETUP & IMPORTS

```python
import pandas as pd
import numpy as np
import yfinance as yf
import pandas_datareader.data as web
from datetime import datetime
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

START_DATE = '2015-01-01'
END_DATE = datetime.today().strftime('%Y-%m-%d')

print(f"Data range: {START_DATE} to {END_DATE}")
```

# STEP 2: FETCH FRED DATA (rates, spreads, real yields)

```python
fred_tickers = {
    'DGS2': 'yield_2y',           # 2Y Treasury - our shock indicator
    'DGS10': 'yield_10y',         # 10Y Treasury
    'DFII10': 'real_yield_10y',   # 10Y TIPS (real yield)
    'DFF': 'fed_funds_rate',      # Effective Fed Funds Rate
    'VIXCLS': 'vix',              # Volatility index
    'BAMLH0A0HYM2': 'hy_spread',  # High Yield credit spread
    'DTWEXBGS': 'dxy_broad',      # Broad Dollar Index (FRED version)
    'CPIAUCSL': 'cpi_index',      # CPI for inflation regime later
}

fred_data = {}
for code, name in fred_tickers.items():
    try:
        series = web.DataReader(code, 'fred', START_DATE, END_DATE)
        fred_data[name] = series[code]
        print(f"✓ Loaded {name} ({len(series)} obs)")
    except Exception as e:
        print(f"✗ Failed to load {name}: {e}")

fred_df = pd.DataFrame(fred_data)
fred_df.index.name = 'date'
```
